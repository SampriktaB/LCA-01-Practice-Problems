```python
#1. Write a program to print the 4th element from first and 4th element from last in a tuple.

Tuple = (10, 20, 30, 40, 50, 60, 70, 80)
print("4th element from first:", Tuple[3])
print("4th element from last:", Tuple[-4])
```

```python
#2. Write a program to check whether an element exists in a tuple or not.

Tuple = (10, 20, 30, 40, 50)
Element = int(input("Enter Element: "))
if Element in Tuple:
    print("Element exists in the tuple.")
else:
    print("Element does not exist in the tuple.")
```

```python
#3. Write a program to convert a list into a tuple. 

List = [10, 20, 30, 40, 50]
Tuple = tuple(List)
print("List:", List)
print("Tuple:", Tuple)
```

```python
#4. Write a program to find the index of an item in a tuple.

Tuple = (10, 20, 30, 40, 50)
Element = int(input("Enter Eement: "))
if Element in Tuple:
    print("Index =", Tuple.index(Element))
else:
    print("Element not found.")
```

```python
#5. Write a program to replace last value of tuples in a list to 100.  
#   Sample list: [(10, 20, 40), (40, 50, 60), (70, 80, 90)]  
#   Expected Output: [(10, 20, 100), (40, 50, 100), (70, 80, 100)] 

List1 = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
List2 = []
for tup in List1:
    Tuple = tup[:-1] + (100,)
    List2.append(Tuple)
print(List2)
```
