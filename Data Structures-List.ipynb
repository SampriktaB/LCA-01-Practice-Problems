```python
#9. Accept 20 values from user and save it in a list. Perform following operations on it:

Numbers = []
for i in range(20):
    Value = int(input("Enter Value: "))
    Numbers.append(Value)
    
#a) Count similar elements of the list and print their index value.

Element = int(input("Enter element to be searched: "))
print("Count:", Numbers.count(Element))
print("Indexes:", end=" ")
for i in range(20):
    if Numbers[i] == Element:
        print(i, end=" ")
print()

#b) Count even and odd values of the list.

Even = 0
Odd = 0
for Value in Numbers:
    if Value % 2 == 0:
        Even += 1
    else:
        Odd += 1
print("Even Numbers:", Even)
print("Odd Numbers:", Odd)

#c) Count positive and negative values of the list.

Positive = 0
Negative = 0
for Value in Numbers:
    if Value > 0:
        Positive += 1
    elif Value < 0:
        Negative += 1
print("Positive Numbers:", Positive)
print("Negative Numbers:", Negative)
```

```python
#10. Accept 10 values from user and save it in a list. Perform following operations on it:

Numbers = []
for i in range(10):
    Value = int(input("Enter Value: "))
    Numbers.append(Value)

#a) Sort the list in ascending order using sorted() function and display the sorted list.

Ascending = sorted(Numbers)
print("Ascending Order:", Ascending)

#b) Sort the list in descending order using sort() function.

Numbers.sort(reverse = True)
print("Descending Order:", Numbers)

#c) Display the length of the list.

print("Length of List:", len(Numbers))
```

```python
#11. Accept two lists from the user and merge them using + in a single list.

List1 = []
List2 = []
print("Enter 5 elements for 1st list:")
for i in range(5):
    Value = int(input("Enter Value: "))
    List1.append(Value)
print("Enter 5 elements for 2nd list:")
for i in range(5):
    Value = int(input("Enter Value: "))
    List2.append(Value)
List3 = List1 + List2
print("Merged List:", List3)
```

```python
#12. An acronym is a word formed by taking the first letters of the words in a phrase and making a word from them. 
#For example, RAM is an acronym for “Random Access Memory.” 
#Write a program that allows the user to type in a phrase and then outputs the acronym for that phrase. 
#Note: The acronym should be all uppercase, even if the words in the phrase are not capitalized.

Phrase = input("Enter Phrase: ")
Words = Phrase.split()
Acronym = ""
for Word in Words:
    Acronym = Acronym + Word[0].upper()
print("Acronym:", Acronym)
```

```python
#13. Write a program to print the abbreviation of a month, given its number.

Month = int(input("Enter Month Number: "))
Months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
if Month >= 1 and Month <= 12:
    print("Month Abbreviation:", Months[Month - 1])
else:
    print("Invalid Month Number")
```
