```python
#1. Write a program to remove a given item from the set.

Set = {10, 20, 30, 40, 50}
Item = int(input("Enter item to be removed: "))
if Item in Set:
    Set.remove(Item)
    print("After removing:", Set)
else:
    print("Item not found.")
```

```python
#2. Write a program to create an intersection of sets.

Set1 = {1, 2, 3, 4, 5}
Set2 = {4, 5, 6, 7, 8}
Intersection = Set1.intersection(Set2)
print("Intersection:", Intersection)
```

```python
#3. Write a program to create a union of sets.

Set1 = {1, 2, 3, 4, 5}
Set2 = {4, 5, 6, 7, 8}
Union = Set1.union(Set2)
print("Union:", Union)
```

```python
#4. Write a program to find the maximum and minimum value in a set.

Set = {10, 20, 5, 40, 30, 15}
Maximum = max(Set)
Minimum = min(Set)
print("Maximum Value:", Maximum)
print("Minimum Value:", Minimum)
```
