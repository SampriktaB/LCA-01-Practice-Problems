```python
#a) Create a Regular Expression and recognize the following strings: “bat,” “bit,” “but,” “hat,” “hit,” or “hut.”

import re as r
String = input("Enter a string: ")
Pattern = r"^(bat|bit|but|hat|hit|hut)$"
if r.match(Pattern, String):
    print("String matched.")
else:
    print("String did not match.")
```

```python
#b) Create a Regular Expression and match any pair of words separated by a single space, i.e., first and last names.

import re as r
String = input("Enter first name and last name: ")
Pattern = r"^[A-Za-z]+ [A-Za-z]+$"
if r.match(Pattern, String):
    print("Valid Name")
else:
    print("Invalid Name")
```

```python
#c) Create a Regular Expression and match any word and single letter separated by a comma and single space, as in last name, first initial.

import re as r
String = input("Enter last name and first initial: ")
Pattern = r"^[A-Za-z]+, [A-Za-z]$"
if r.match(Pattern, String):
    print("Valid")
else:
    print("Invalid")
```
