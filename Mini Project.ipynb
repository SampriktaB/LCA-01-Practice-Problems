```python
#1. Create a python program that asks the user how far they want to travel. 
#If they want to travel less than 3 miles tell them to ride a bicycle, 
#if they want to travel more than 3 miles but less than 300 miles tell them to ride a motorcycle and,
#if they want to travel 300 miles or more tell them to driver supercar.

Distance = float(input("Enter the distance you want to travel in miles: "))
if Distance < 3:
    print("You should ride a bicycle.")
elif Distance < 300:
    print("You should ride a motorcycle.")
else:
    print("You should travel by a supercar.")
```

```python
#2. Let's assume you are planning to use your python skills to build PBLApp for mobile.
#You decide to host your application on servers running in the cloud. You pick a hosting provider that charges $0.51 per hour. 
#You will launch your service using one server and want to know how much it will cost to operate per day, per week and, per month. 
#Write a python code that displays answers of following questions: 
#How much does it costs to operate one server per day? 
#How much does it costs to operate one server per week? 
#How much does it costs to operate one server per month? 
#How many days can I operate one server with $918?

Hourly_Cost = 0.51
Per_Day = Hourly_Cost * 24
Per_Week = Per_Day * 7
Per_Month = Per_Day * 30
Days = 918 / Per_Day
print("Cost per day = $", Per_Day)
print("Cost per week = $", Per_Week)
print("Cost per month = $", Per_Month)
print("Number of days for $918 =", Days)
```
