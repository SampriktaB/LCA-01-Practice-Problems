```python
#1. Write a program to count the number of upper and lower case letters in a string.

String = input("Enter a string: ")
Upper = 0
Lower = 0
for ch in String:
    if ch.isupper():
        Upper += 1
    elif ch.islower():
        Lower += 1
print("Number of uppercase letters:", Upper)
print("Number of lowercase letters:", Lower)
```

```python
#2. Write a program that will check whether a given string is Palindrome or not.

String = input("Enter a string: ")
if String == String[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

```python
#3. Given a string, return a new string made of n copies of the first 2 characters of the original string where n is the length of the string. 
#The string length will be >=2. If the input is "Wipro", then output should be "Wiwiwiwiwi". 

String = input("Enter a string: ")
n = int(input("Enter n: "))
if len(String) >= 2:
    New_String = String[:2] * n
    print("New String:", New_String)
else:
    print("String length must be at least 2.")
```

```python
#4. Given a string, if the first or last character is 'x', return the string without those 'x' character, else return the string unchanged. 
#If the input is "xHix", then output should be "Hi".

String = input("Enter a string: ")
if String.startswith('x'):
    String = String[1:]
if String.endswith('x'):
    String = String[:-1]
print("Result:", String)
```

```python
#5. Given a string and an integer n, return a string made of n repetitions of the last n characters of the string. 
#You may assume that n is between 0 and the length of the string inclusive. 
#For example if the inputs are "Wipro" and 3, then the output should be "propropro". 

String = input("Enter a string: ")
n = int(input("Enter n: "))
if 0 <= n <= len(String):
    Result = String[-n:] * n
    print("Result:", Result)
else:
    print("Invalid value of n.")
```
