```python
#1. Write a program to add a key and value to a dictionary.  
#Sample Dictionary: (0:10, 1: 20)  
#Expected Result: (0: 10, 1:20, 2:30) 

Dict = {0: 10, 1: 20}
Dict[2] = 30
print(Dict)
```

```python
#2. Write a program to concatenate the following dictionaries to create a new one.  
#     Sample Dictionary:  
#           dic1=(1:10, 2:20) dic2={3:30, 4:40) dic3={5:50,6:60}  
#     Expected Result (1: 10, 2: 20, 3: 30, 4: 40, 5:50, 6: 60) 

dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}
dict4 = {}
dict4.update(dic1)
dict4.update(dic2)
dict4.update(dic3)
print(dict4)
```

```python
#3. Write a program to check if a given key already exists in a dictionary.

Dict = {1: 10, 2: 20, 3: 30, 4: 40}
Key = int(input("Enter the key to search: "))
if Key in Dict:
    print("Key already exists.")
else:
    print("Key does not exist.")
```

```python
#4. Write a program to iterate over a dictionary using for loop and print the keys alone, values alone and both keys and values.

Dict = {1: 10, 2: 20, 3: 30, 4: 40}
print("Keys:")
for Key in Dict.keys():
    print(Key)
print("Values:")
for Value in Dict.values():
    print(Value)
print("Both keys and values:")
for Key, Value in Dict.items():
    print(Key, Value)
```

```python
#5. Write a program to prepare a dictionary where the keys are numbers between 1 and 15 (both included) and the values are square of the keys.

Dict = {}
for i in range(1, 16):
    Dict[i] = i * i
print(Dict)
```

```python
#6. Write a program to sum all the values in a dictionary, considering the values will be of int type.

Dict = {1: 10, 2: 20, 3: 30, 4: 40}
Total = sum(Dict.values())
print("Sum of values =", Total)
```

```python
#8.  Convert list to dictionary.
#    e.g. list1=[‘name’, ‘panel’,’rollno’] 
#         list2=[‘ABC’,’B’,34]  
#         my_dict= { ‘name’:’ABC’, ‘panel’=’B’, ‘rollno’=34}

list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]
my_dict = dict(zip(list1, list2))
print(my_dict)
```

```python
#9. Convert dictionary to lists. Make one list of keys and other of values.

Dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34
}
Keys = list(Dict.keys())
Values = list(Dict.values())
print("Keys:", Keys)
print("Values:", Values)
```

```python
#10. Find the mean of values of dictionary.
#    e.g mydict= { ‘marks1’: 23, ‘marks2’: 123, ‘marks3’: 43, ‘marks4’: 13, ‘marks5’: 39) 
#    output: mean of all values.

my_dict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}
Total = sum(my_dict.values())
Mean = Total / len(my_dict)
print("Mean of all values =", Mean)
```

```python
#11. Sort the elements of the dictionary using sorted() function.
#    e.g my_dict= { ‘name’:’ABC’, ‘panel’=’B’, ‘rollno’=34, ‘marks’=[65,87,67,94]} 
#    sorted(my_dict)

my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}
Sorted_Dict = dict(sorted(my_dict.items()))
print(Sorted_Dict)
```
